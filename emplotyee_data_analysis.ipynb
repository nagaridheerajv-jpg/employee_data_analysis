```python
# Task - 1
# Five Visualisations
# Requirements
# • Create at least 5 charts: bar, histogram, scatter, line (time trend or sorted values) and one
# more of your choice (box plot or heatmap)
# • Every chart needs a title, labelled axes, units and a legend where needed
# Add one sentence under each chart describing the insight
# Saving figures to reports/figures/ with a consistent color scheme.

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure output directory exists
os.makedirs('reports/figures', exist_ok=True)

# Load the dataset
df = pd.read_csv('/content/clean_data.csv')

# Define consistent color scheme
PRIMARY_COLOR = '#1f77b4'  # Steel blue
ACCENT_COLOR = '#ff7f0e'   # Muted orange
GRID_COLOR = '#e5e5e5'
sns.set_theme(style='whitegrid', palette='deep')
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.color'] = GRID_COLOR

# 1. BAR CHART: Employees by Department
dept_count = df["department"].value_counts().sort_index()

plt.figure(figsize=(7, 4.5))
plt.bar(dept_count.index, dept_count.values,
        label="Employees", color=PRIMARY_COLOR, edgecolor='none')

plt.title("Employee Count by Department", fontsize=12, fontweight='bold')
plt.xlabel("Department", fontsize=10)
plt.ylabel("Number of Employees (count)", fontsize=10)
plt.legend()
plt.figtext(
    0.5, 0.01,
    "Insight: IT has the highest employee count with 3 employees.",
    ha="center", wrap=True, fontsize=10, fontstyle='italic'
)
plt.tight_layout()
plt.savefig('reports/figures/1_employee_count_by_department.png', dpi=300)
plt.show()


# 2. HISTOGRAM: Salary Distribution
plt.figure(figsize=(7, 4.5))
plt.hist(df["salary"], bins=5, edgecolor="white",
         label="Employees", color=PRIMARY_COLOR)

plt.title("Employee Salary Distribution", fontsize=12, fontweight='bold')
plt.xlabel("Salary (INR per month)", fontsize=10)
plt.ylabel("Number of Employees (count)", fontsize=10)
plt.legend()
plt.figtext(
    0.5, 0.01,
    "Insight: Salaries range from ₹30,000 to ₹90,000 per month.",
    ha="center", wrap=True, fontsize=10, fontstyle='italic'
)
plt.tight_layout()
plt.savefig('reports/figures/2_employee_salary_distribution.png', dpi=300)
plt.show()


# 3. SCATTER PLOT: Employee ID vs Salary
plt.figure(figsize=(7, 4.5))
plt.scatter(df["id"], df["salary"],
            label="Employees", color=PRIMARY_COLOR, s=80, edgecolors='none')

plt.title("Employee ID vs Monthly Salary", fontsize=12, fontweight='bold')
plt.xlabel("Employee ID (identifier)", fontsize=10)
plt.ylabel("Salary (INR per month)", fontsize=10)
plt.legend()
plt.figtext(
    0.5, 0.01,
    "Insight: Employee IDs are identifiers, so they do not establish a meaningful salary relationship.",
    ha="center", wrap=True, fontsize=10, fontstyle='italic'
)
plt.tight_layout()
plt.savefig('reports/figures/3_employee_id_vs_salary.png', dpi=300)
plt.show()


# 4. LINE CHART: Sorted Salary Values
sorted_salary = df["salary"].sort_values().reset_index(drop=True)

plt.figure(figsize=(7, 4.5))
plt.plot(
    range(1, len(sorted_salary) + 1),
    sorted_salary,
    marker="o",
    markersize=8,
    label="Salary",
    color=PRIMARY_COLOR
)

plt.title("Sorted Employee Salaries", fontsize=12, fontweight='bold')
plt.xlabel("Salary Rank (lowest to highest)", fontsize=10)
plt.ylabel("Salary (INR per month)", fontsize=10)
plt.legend()
plt.figtext(
    0.5, 0.01,
    "Insight: The lowest salary is ₹30,000 and the highest is ₹90,000 per month.",
    ha="center", wrap=True, fontsize=10, fontstyle='italic'
)
plt.tight_layout()
plt.savefig('reports/figures/4_sorted_employee_salaries.png', dpi=300)
plt.show()


# 5. BOX PLOT: Salary Distribution
plt.figure(figsize=(7, 4.5))
sns.boxplot(y=df["salary"], color=ACCENT_COLOR, width=0.4)

plt.title("Box Plot of Employee Salaries", fontsize=12, fontweight='bold')
plt.xlabel("Salary Distribution", fontsize=10)
plt.ylabel("Salary (INR per month)", fontsize=10)
plt.figtext(
    0.5, 0.01,
    "Insight: The median salary is ₹56,000, and no outliers are flagged by the IQR rule.",
    ha="center", wrap=True, fontsize=10, fontstyle='italic'
)
plt.tight_layout()
plt.savefig('reports/figures/5_boxplot_employee_salaries.png', dpi=300)
plt.show()
```
